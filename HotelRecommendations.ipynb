```python
def map(x):
  x =str(x).lower()
  if any(i in x for i in ['farm stay','motel','hostel','tent']):
    return 'Bnb'
  else:
    return x.title()

data['property_type'] = data['property_type'].apply(map)
```

```python
#creating a random price column based on roomtype
def price_gen(property_type,room_type):
  base = {
      'Hotel':3500,
      'Resort':6000,
      'Homestay':2800,
      'Guest House':2500,
      'Service Apartment':3000,
      'Lodge':1500,
      'Bnb':2000,
      'Houseboat':4000,
      'Cottage':4500,
      'Villa':8000,
      'Bungalow':10000,
      'Palace':15000
  }

  extra = {
      'Standard': 0,
      'Deluxe': 300,
      'Superior': 500,
      'Executive': 750,
      'Suite': 1500,
      'Luxury': 1000
  }

  base_price = base.get(property_type,3000)
  room_charges = extra.get(room_type,1000)
  noise = np.random.randint(-500,2000)
  final_price = base_price + room_charges + noise
  return final_price
```

```python
data['price'] = data.apply(
    lambda x: price_gen(x['property_type'],
                        x['room_type_cleaned']),axis=1)
```
